> ### part I
> * 將 下列圖片 彩色載入
> * 垂直 rotate 90 度, show 出結果
> * 再 3x3 高斯濾波器, sigmaX=0.8, sigmaY=0.8, show 出結果<br>

> ### part II
> * 再 sobelX, sobelY 後用(0.5, 0.5)的 權重加起來, show 出結果<br>
> * 再輪廓偵測外框 (contours) 將找到的輪廓填滿畫在原圖上, show 出結果
> * 將上圖中的物件`中心`用圓圈標示出來, 並將`面積`標示於中心旁邊

### 必須將上列個各處理後的圖 imshow 出來

```python
import cv2
import numpy as np

img = cv2.imread('../image/contour.png')

# -------- 垂直 rotate 90 度 ------------------------
# cv2.imshow('orginal', img)
img90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
# cv2.imshow('Rotate 90', img90)


# ----- 3*3 高斯濾波器 -----------------------------
sigma =  cv2.GaussianBlur(img90 , (3,3), 0.8, 0.8)
# cv2.imshow('sigma0', sigma)  

# ------- 將前一處理後圖片取 SobelX, SobelY and addWeight(0.5, 0.5)----------

sobelx = cv2.Sobel(sigma, cv2.CV_64F, 1, 0)   # ddepth = cv2.CV64F
sobely = cv2.Sobel(sigma, cv2.CV_64F, 0, 1)   # ddepth = cv2.CV64F

sobelx = cv2.convertScaleAbs(sobelx)        # 絕對值, 轉換為cv2.CV_8U  
sobely = cv2.convertScaleAbs(sobely)        # 絕對值, 轉換為cv2.CV_8U  
sobelxy = cv2.addWeighted(sobelx, 0.5, sobely, 0.5, 0) 

# cv2.imshow('sobelxy', sobelxy)

# ------- 輪廓偵測 (contours) 將找到的輪廓畫在原圖上 ----------

cv2.imshow('original', img)
imgray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(imgray, 127, 255, cv2.THRESH_BINARY_INV)
print(ret)






# ------- 將上圖中的物件中心用圓圈標示出來, 並將面積標示於中心旁邊 ---------






cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```
